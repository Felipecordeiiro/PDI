```python
import numpy as np
import cv2 as cv
import os

def remontar_image_from_bits(bits_para_usar, diretorio_planos, caminho_saida="planos_de_bits"):
    """
    Remonta uma imagem a partir de um conjunto selecionado de planos de bits usando Python.

    Args:
        bits_para_usar (list): Lista de inteiros especificando quais bits (0 a 7) usar.
        diretorio_planos (str): Diretório onde os arquivos plano_bit_*.png estão.
        caminho_saida (str): Onde salvar a imagem reconstruída.
    """
    print(f"--- Remontando imagem com Python usando os bits: {bits_para_usar} ---")

    planos_carregados = []
    tamanho_imagem = None
    for bit in bits_para_usar:
        try:
            caminho_bit = os.path.join(diretorio_planos, f"plano_bit_{bit}.png")
            img_bit = cv.imread(caminho_bit, cv.IMREAD_GRAYSCALE)
            planos_carregados.append(img_bit)
            
            if tamanho_imagem is None:
                tamanho_imagem = img_bit.shape 

        except Exception as e:
            print(f"Ocorreu um erro inesperado: {e}")
    
    if not planos_carregados:
        print("Nenhum bit selecionado. Saindo.")
        return
    
    imagem_remontada_temp = np.zeros(tamanho_imagem, dtype=np.uint16)

    for i, plano_array in enumerate(planos_carregados):
        if plano_array.shape != tamanho_imagem:
            print(f"Erro de dimensão! O plano do bit {bits_para_usar[i]} tem tamanho {plano_array.shape} mas o esperado era {tamanho_imagem}")
            return
        
        bit_atual = bits_para_usar[i]
        
        # Calcula o valor da potência de 2 (ex: 2^7 = 128)
        valor_bit = 2**bit_atual
        
        # Onde o pixel no plano de bit for branco (>0), adicionamos o valor_bit
        # à nossa imagem final. Esta é uma operação vetorizada do NumPy, muito rápida!
        imagem_remontada_temp[plano_array > 0] += valor_bit

    # Converte o resultado final de volta para 8 bits
    imagem_remontada = imagem_remontada_temp.astype(np.uint8)
    try:
        cv.imwrite(caminho_saida+"lalala.png", imagem_remontada)
        print(f"Imagem remontada com sucesso e salva em: {caminho_saida}\n")
    except Exception as e:
        print(f"Erro ao salvar a imagem de saida: {e}")
```

```python
path_image_bits = "./../3.14/planos_de_bits/"
bits_for_use = [5, 6, 7]
remontar_image_from_bits(bits_for_use, path_image_bits)
```
